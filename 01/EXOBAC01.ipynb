Question 2

```python
def initialiser(n) : 
    resu = []
    for i in range(n) : 
        resu.append(False)
    return resu
```

```python
initialiser(3)
```

```python
def initialiser(n) : 
    return  [False for i in range(n) ]
```

```python
initialiser(4)
```

```python
def victoire(tab):
    for etat_case in tab : 
        print(etat_case)
        if etat_case == False : 
            return False
    return True
```

```python
victoire( [True, True, True] )
```

```python
victoire( [True, True, False] )
```

```python
victoire( [True, True, False, True, True] )
```
